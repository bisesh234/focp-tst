```python
def all_unique(nums):
    return len(nums) == len(set(nums))

print(all_unique([1, 2, 3, 4]))
print(all_unique([1, 2, 3, 3]))
```

```python
longest = ""

while True:
    s = input("Enter a string: ")
    if s == "":
        break
    if len(s) > len(longest):
        longest = s

print("Longest string:", longest)
```
